# orders_dataset

**This dataset contains 99441 rows and 8 columns.**

**Dataset contains 4908 missing values in total.**

## columns_overview

1.> **order_id** -> This column contains the ID associated with each order placed by a customer. The datatype is string and it contains 99441 unique values.

2.> **customer_id** -> This column contains the ID associated with the customer who placed the order. The datatype is string and it contains 99441 unique values.

3.> **order_status** -> This column represents the current/final status of an order, such as delivered, canceled, shipped, processing, etc. The datatype is string.

4.> **order_purchase_timestamp** -> This column contains the date and time when the customer placed the order. The datatype is string, whereas it should be datetime.

5.> **order_approved_at** -> This column contains the date and time when the order was approved. The datatype is string, whereas it should be datetime.

**This column contains 160 missing values.**

6.> **order_delivered_carrier_date** -> This column contains the date and time when the order was handed over to the delivery carrier. The datatype is string, whereas it should be datetime.

**This column contains 1783 missing values.**

7.> **order_delivered_customer_date** -> This column contains the date and time when the order was delivered to the customer. The datatype is string, whereas it should be datetime.

**This column contains 2965 missing values.**

8.> **order_estimated_delivery_date** -> This column contains the estimated date and time of delivery of the order. The datatype is string, whereas it should be datetime.

## Important observations

1.> Missing values are not randomly distributed across different order statuses.

2.> Most missing delivery timestamps are associated with canceled, shipped, invoiced, processing and unavailable orders.

3.> 1107 shipped orders have missing customer delivery timestamps, which is understandable because shipped orders are not necessarily delivered yet.

4.> 8 delivered orders have missing customer delivery timestamps, and 14 delivered orders have missing approval timestamps. These records require further investigation.

5.> All 609 unavailable orders have missing carrier handover and customer delivery timestamps.

**We will not remove or replace these missing values during the assessment stage. Their treatment will be decided after further investigation.**
